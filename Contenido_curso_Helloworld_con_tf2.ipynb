## *Justificación:*
El avance en inteligencia artificial ha revolucionado la capacidad de analizar datos y extraer patrones complejos mediante el  aprendizaje profundo. Deep Learning I proporciona una comprensión fundamental de las técnicas de aprendizaje supervisado  y no supervisado, esenciales para el desarrollo de modelos adaptativos y autónomos. Los estudiantes desarrollarán  competencias para diseñar e implementar soluciones innovadoras en ciencia de datos e inteligencia artificial. Esta asignatura 
fomenta el pensamiento crítico y la experimentación en escenarios complejos, promoviendo la aplicación ética y estratégica de  estas metodologías en la industria, la investigación y la academia.



### *REQUISITOS ACADÉMICOS*:
Personas con conocimientos en programación Python y conocimientos básicos en
álgebra lineal, que quieran incluir TensorFlow en su caja de herramientas para la
aplicación de modelos supervisados de Deep Learning. Además, el estudiante debe tener
ya sea una cuenta en Google para acceder a processos de GPU a través de Colab
notebook, o un computador con los requerimientos básicos para la instalación de
TensorFlow con o sin GPU (https://www.tensorflow.org/install).
